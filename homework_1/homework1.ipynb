```python
asia = df[df['origin'] == 'Asia']
```

```python
X = asia[['vehicle_weight', 'model_year']].head(7).values
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_INVERT = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_INVERT @ X.T @ y
w
```
